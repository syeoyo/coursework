```python
import gurobipy as gp
from gurobipy import GRB

# 1. 모델 생성
model = gp.Model("Primal_LP")

# 2. 변수 정의 (x1, x2, x3, x4 >= 0)
x1 = model.addVar(vtype=GRB.CONTINUOUS, lb=0, name="x1")
x2 = model.addVar(vtype=GRB.CONTINUOUS, lb=0, name="x2")
x3 = model.addVar(vtype=GRB.CONTINUOUS, lb=0, name="x3")

# 3. 목적 함수 정의 (Minimize Z = 2x1 + 15x2 + 5x3 + 6x4)
model.setObjective(2*x1 + 3*x2 + 4*x3, GRB.MINIMIZE)

# 4. 제약 조건 추가
model.addConstr(x1 + 2*x2 + x3 >= 3, "C1")
model.addConstr(2*x1 - x2 + 3*x3 >= 4, "C2")

# 5. 모델 최적화 실행
model.optimize()

# 6. 결과 출력
if model.status == GRB.OPTIMAL:
    print("\n--- Optimal Solution (Primal) ---")
    print(f"Optimal Objective Value (Z): {model.objVal}")
    print(f"x1: {x1.x}")
    print(f"x2: {x2.x}")
    print(f"x3: {x3.x}")

    # 쌍대 변수 값 (Dual Values, Shadow Prices) 추출
    # Gurobi는 제약 조건에 대한 .pi 속성을 통해 쌍대 변수 값을 제공합니다.
    c1_constr = model.getConstrByName("C1")
    c2_constr = model.getConstrByName("C2")
    
    # Primal C1 (>=)의 쌍대 변수는 w1
    # Primal C2 (<=)의 쌍대 변수는 w2 (Gurobi의 쌍대 변수 부호는 Maximize-type Dual 기준)
    print("\n--- Dual Variables (w*) ---")
    print(f"w1 (Dual var for C1): {c1_constr.pi}")
    print(f"w2 (Dual var for C2): {c2_constr.pi}")
    
else:
    print("Optimization was not successful.")
```

```python
import gurobipy as gp
from gurobipy import GRB

# 1. 모델 생성
model = gp.Model("Primal_LP")

# 2. 변수 정의 (x1, x2, x3, x4 >= 0)
x1 = model.addVar(vtype=GRB.CONTINUOUS, name="x1")
x2 = model.addVar(vtype=GRB.CONTINUOUS, name="x2")
x3 = model.addVar(vtype=GRB.CONTINUOUS, name="x3")
x4 = model.addVar(vtype=GRB.CONTINUOUS, name="x4")

# 3. 목적 함수 정의 (Minimize Z = 2x1 + 15x2 + 5x3 + 6x4)
model.setObjective(2*x1 + 15*x2 + 5*x3 + 6*x4, GRB.MINIMIZE)

# 4. 제약 조건 추가
model.addConstr(x1 + 6*x2 + 3*x3 + x4 >= 2, "C1")
model.addConstr(-2*x1 + 5*x2 - 4*x3 + 3*x4 <= -3, "C2")

# 5. 모델 최적화 실행
model.optimize()

# 6. 결과 출력
if model.status == GRB.OPTIMAL:
    print("\n--- Optimal Solution (Primal) ---")
    print(f"Optimal Objective Value (Z): {model.objVal}")
    print(f"x1: {x1.x}")
    print(f"x2: {x2.x}")
    print(f"x3: {x3.x}")
    print(f"x4: {x4.x}")

    # 쌍대 변수 값 (Dual Values, Shadow Prices) 추출
    # Gurobi는 제약 조건에 대한 .pi 속성을 통해 쌍대 변수 값을 제공합니다.
    c1_constr = model.getConstrByName("C1")
    c2_constr = model.getConstrByName("C2")
    
    # Primal C1 (>=)의 쌍대 변수는 w1
    # Primal C2 (<=)의 쌍대 변수는 w2 (Gurobi의 쌍대 변수 부호는 Maximize-type Dual 기준)
    print("\n--- Dual Variables (w*) ---")
    print(f"w1 (Dual var for C1): {c1_constr.pi}")
    print(f"w2 (Dual var for C2): {c2_constr.pi}")
    
else:
    print("Optimization was not successful.")
```

```python
import gurobipy as gp
from gurobipy import GRB

# 1. 모델 생성
model_dual = gp.Model("Dual_LP")

# 2. 변수 정의 (w1, w2 >= 0)
w1 = model_dual.addVar(vtype=GRB.CONTINUOUS, name="w1")
w2 = model_dual.addVar(vtype=GRB.CONTINUOUS, name="w2")

# 3. 목적 함수 정의 (Maximize W = 2w1 + 3w2)
model_dual.setObjective(2*w1 + 3*w2, GRB.MAXIMIZE)

# 4. 제약 조건 추가
model_dual.addConstr(w1 + 2*w2 <= 2, "D1")
model_dual.addConstr(6*w1 - 5*w2 <= 15, "D2")
model_dual.addConstr(3*w1 + 4*w2 <= 5, "D3")
model_dual.addConstr(w1 - 3*w2 <= 6, "D4")

# 5. 모델 최적화 실행
model_dual.optimize()

# 6. 결과 출력
if model_dual.status == GRB.OPTIMAL:
    print("\n--- Optimal Solution (Dual) ---")
    print(f"Optimal Objective Value (W): {model_dual.objVal}")
    print(f"w1: {w1.x}")
    print(f"w2: {w2.x}")

    # 원시 변수 값 (Primal Variables, Reduced Costs) 추출
    # Gurobi는 변수에 대한 .rc 속성을 통해 원시 변수 값을 제공합니다.
    # 단, 목적 함수가 Maximize일 때 .rc는 (cj - w*aj)에 해당합니다. 
    # 원시 문제의 해 (x*)는 쌍대 문제의 제약 조건에 대한 'Reduced Cost'를 통해 얻을 수 있습니다.
    # Gurobi가 내부적으로 쌍대 관계를 처리하므로, 쌍대 변수의 reduced cost는 원시 변수의 최적 해가 됩니다.
    print("\n--- Primal Variables (x*) from Dual ---")
    print(f"x1 (RC for D1): {model_dual.getConstrByName('D1').pi}")
    print(f"x2 (RC for D2): {model_dual.getConstrByName('D2').pi}")
    print(f"x3 (RC for D3): {model_dual.getConstrByName('D3').pi}")
    print(f"x4 (RC for D4): {model_dual.getConstrByName('D4').pi}")
    
else:
    print("Optimization was not successful.")
```
